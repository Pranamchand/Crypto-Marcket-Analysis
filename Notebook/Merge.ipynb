```python
# Import Libreries
from pathlib import Path
import pandas as pd

# Project root directory
ROOT_DIR = Path.cwd().parent

# Raw Dataset 
DATASET_PATH = ROOT_DIR / "Datasets" / "row"

files = list(DATASET_PATH.glob("coin_*.csv"))

print("Number of CSV files found:", len(files))

for file in files:
    print(file.name)
print("\nCurrent Path", Path.cwd())
```

```python
dataframes = []

for file in files:

    df = pd.read_csv(file)

    crypto_name = file.stem.replace("coin_", "")

    df["Crypto"] = crypto_name

    dataframes.append(df)

master_df = pd.concat(dataframes, ignore_index=True)

master_df.head()
```

```python
print("Total rows:", len(master_df))
print("Total columns:", len(master_df.columns))
print("Number of cryptocurrencies:", master_df["Crypto"].nunique())

print("\nCryptocurrencies:")
print(master_df["Crypto"].unique())

print("\nColumns:")
print(master_df.columns.tolist())
```

```python
# Convert Datetime data type
master_df["Date"] = pd.to_datetime(
    master_df["Date"],
    format="mixed"
)

# Sort by cryptocurrency and date
master_df = master_df.sort_values(
    by=["Crypto", "Date"]
).reset_index(drop=True)

master_df.head(10)
```

```python
MERGED_DATASET = ROOT_DIR / "Datasets"  / "merged" / "crypto_master.csv"

master_df.to_csv(MERGED_DATASET, index=False)
print("Merged dataset saved successfully!")
print(MERGED_DATASET)
```
