```python
# Import Dataset
from pathlib import Path
import pandas as pd

# Project root directory
ROOT_DIR = Path.cwd().parent

# Merged Dataset 
DATASET_PATH = ROOT_DIR / "Datasets"  / "merged" / "crypto_master.csv"

df = pd.read_csv(DATASET_PATH)
df.head()
```

```python
df.info()
```

```python
# TO Date datatype
df['Date'] = pd.to_datetime(df['Date'])
```

instead of deleting the zero volume and marketcap columns I am going to create a new column named is_zero_volume and is_zero_marketcap. 

```python
# Create boolean columns for zero volume and zero market cap
df['is_zero_volume'] = df['Volume'] == 0
df['is_zero_marketcap'] = df['Marketcap'] == 0
```

```python
# Delete serial number
df = df.drop(columns = ['SNo'])
```

```python
df.head()
```

For now let's ignore the date inconsistency I beleve it doesn't gonna effect our analysis. 

```python
CLEANED_DATASET_PATH = ROOT_DIR / "Datasets" / "cleaned" / "crypto_master_cleaned.csv"

df.to_csv(CLEANED_DATASET_PATH, index=False)
print("dataset updated successfully.")
```
